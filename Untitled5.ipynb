```python
#CATEGORY 1 = 1.Simple Reflex Agent - Vacuum Cleaner

def simple_reflex_agent(location, status):
    """
    location: 'A' or 'B'
    status: 'Dirty' or 'Clean'
    """

    if status == "Dirty":
        return "Suck"
    elif location == "A":
        return "Move Right"
    elif location == "B":
        return "Move Left"

print("Location: A, Status: Dirty ->", simple_reflex_agent("A", "Dirty"))
print("Location: A, Status: Clean ->", simple_reflex_agent("A", "Clean"))
print("Location: B, Status: Dirty ->", simple_reflex_agent("B", "Dirty"))
print("Location: B, Status: Clean ->", simple_reflex_agent("B", "Clean"))
```

```python
#2.Model-Based Agent - Door Security System

class DoorSecurityAgent:
    def __init__(self):
        self.last_person = None

    def update_model(self, person):
        self.last_person = person
    def act(self, person, authorized):
        self.update_model(person)
        if authorized:
            return "Door Opened"
        else:
            return "Access Denied"

agent = DoorSecurityAgent()

people = [
    ("Alice", True),
    ("Bob", False),
    ("Charlie", True),
    ("David", False)
]

for person, authorized in people:
    action = agent.act(person, authorized)

    print("Person:", person)
    print("Authorized:", authorized)
    print("Action:", action)
    print("Last Person Stored:", agent.last_person)
    print("-" * 35)
```

```python
#3.Goal-Based Agent - Water Bottle Filling

class WaterBottleAgent:
    def __init__(self):
        self.goal = 100

    def act(self, level):

        if level >= self.goal:
            return "Bottle Full"
        else:
            return "Fill Water"

agent = WaterBottleAgent()
levels = [20, 50, 80, 100]
for level in levels:
    print("Water Level:", level, "%")
    print("Goal:", agent.goal, "%")
    print("Action:", agent.act(level))
    print("-" * 30)
```

```python
#4.Utility-Based Agent - Air Conditioner

class UtilityAgent:
    def utility(self, temperature):

        if temperature == 24:
            return 10

        elif temperature == 22:
            return 8

        elif temperature == 26:
            return 6

        else:
            return 2
    def choose_best(self, temperatures):

        best = temperatures[0]

        for temp in temperatures:
            if self.utility(temp) > self.utility(best):
                best = temp

        return best

agent = UtilityAgent()

temperatures = [22, 24, 26]

best_temperature = agent.choose_best(temperatures)

print("Available Temperatures:", temperatures)
print("Best Temperature:", best_temperature)
print("Utility Value:", agent.utility(best_temperature))
```

```python
#CATEGORY 2 = 1.Fully Observable Agent
class FullyObservableAgent:
    def __init__(self, grid):
        self.grid = grid

    def find_goal(self):
        for i in range(len(self.grid)):
            for j in range(len(self.grid[0])):
                if self.grid[i][j] == "G":
                    return (i, j)

grid = [
    ["S", ".", "."],
    [".", ".", "G"],
    [".", ".", "."]
]

agent = FullyObservableAgent(grid)
print("Goal found at:", agent.find_goal())
```

```python
#1.Partially Observable Agent
class PartiallyObservableAgent:
    def __init__(self, grid):
        self.grid = grid
        self.position = (0, 0)

    def observe(self):
        x, y = self.position
        print("Current Cell:", self.grid[x][y])

    def move(self, direction):
        x, y = self.position

        if direction == "right":
            y += 1
        elif direction == "left":
            y -= 1
        elif direction == "up":
            x -= 1
        elif direction == "down":
            x += 1

        if 0 <= x < len(self.grid) and 0 <= y < len(self.grid[0]):
            self.position = (x, y)

grid = [
    ["S", ".", "."],
    [".", ".", "G"],
    [".", ".", "."]
]

agent = PartiallyObservableAgent(grid)

agent.observe()
agent.move("right")
agent.observe()
agent.move("down")
agent.observe()
```

```python
#2.Known Environment
class KnownEnvironmentAgent:
    def shortest_path(self):
        return ["Right", "Right", "Down"]

agent = KnownEnvironmentAgent()

print("Planned Path:")
print(agent.shortest_path())
```

```python
#2.Unknown Environment
import random

class UnknownEnvironmentAgent:
    def explore(self):
        actions = ["Left", "Right", "Up", "Down"]
        action = random.choice(actions)
        print("Exploring:", action)

agent = UnknownEnvironmentAgent()

for _ in range(5):
    agent.explore()
```

```python
#3.Static Environment
class StaticAgent:
    def perceive(self):
        print("Environment remains unchanged.")

agent = StaticAgent()
agent.perceive()
```

```python
#3.Dynamic Environment
import random
import time

class DynamicEnvironment:
    def __init__(self):
        self.temperature = 25

    def update(self):
        self.temperature += random.randint(-2, 2)

env = DynamicEnvironment()

for i in range(5):
    env.update()
    print("Current Temperature:", env.temperature)
    time.sleep(1)
```

```python
#4.Discrete Environment
class DiscreteAgent:
    def move(self, state):
        transitions = {
            "A": "B",
            "B": "C",
            "C": "Goal"
        }
        return transitions.get(state, state)

agent = DiscreteAgent()

state = "A"

while state != "Goal":
    print(state)
    state = agent.move(state)

print("Reached Goal")
```

```python
#4.Continuous Environment
class ContinuousAgent:
    def __init__(self):
        self.position = 0.0

    def move(self, distance):
        self.position += distance
        print("Position:", self.position)

agent = ContinuousAgent()

agent.move(0.8)
agent.move(1.2)
agent.move(2.5)
```

```python
#5.Deterministic Environment
class DeterministicAgent:
    def move(self):
        print("Moved Right")

agent = DeterministicAgent()

for _ in range(5):
    agent.move()
```

```python
#5.Stochastic Environment
import random

class StochasticAgent:
    def move(self):
        actions = ["Left", "Right", "Up", "Down"]
        print("Moved", random.choice(actions))

agent = StochasticAgent()

for _ in range(5):
    agent.move()
```

```python
#6.Episodic Environment
class EpisodicAgent:
    def classify(self, image):
        if image == "cat":
            return "Cat"
        else:
            return "Dog"

agent = EpisodicAgent()

images = ["cat", "dog", "cat"]

for img in images:
    print(img, "->", agent.classify(img))
```

```python
#6.Sequential Environment
class SequentialAgent:
    def __init__(self):
        self.score = 0

    def move(self, reward):
        self.score += reward
        print("Current Score:", self.score)

agent = SequentialAgent()

agent.move(10)
agent.move(-5)
agent.move(20)
```
