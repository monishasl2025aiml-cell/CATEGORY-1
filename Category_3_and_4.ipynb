```python
#Category-3  1.Exploration–Exploitation trade-off
import random
states = [10, 25, 40, 60, 55, 50, 70, 85, 100]

print("Search Space:", states)
print("\n===== LOCAL SEARCH (Exploitation) =====")

current = 0
print("Start State:", current, "Value:", states[current])

while current < len(states) - 1:
    if states[current + 1] > states[current]:
        current += 1
        print("Move to State", current, "Value:", states[current])
    else:
        break

print("Local Optimum Found")
print("State:", current)
print("Value:", states[current])

print("\n===== GLOBAL SEARCH (Exploration) =====")

best_state = 0
best_value = states[0]

for i in range(5):
    random_state = random.randint(0, len(states) - 1)

    print("Random State:", random_state,
          "Value:", states[random_state])

    if states[random_state] > best_value:
        best_state = random_state
        best_value = states[random_state]

print("\nGlobal Optimum Found")
print("State:", best_state)
print("Value:", best_value)
```

```python
#2.Hill Climbing Search
from collections import deque
def heuristic(state, goal):
    return abs(goal - state)

def hill_climbing(start, goal):
    current = start
    path = [current]

    while current != goal:
        neighbors = [current - 1, current + 1]

        best_neighbor = min(neighbors, key=lambda x: heuristic(x, goal))

        if heuristic(best_neighbor, goal) >= heuristic(current, goal):
            print("\nHill Climbing (Local Search):")
            print("Path:", path)
            print("Stopped at local optimum")
            return

        current = best_neighbor
        path.append(current)

    print("\nHill Climbing (Local Search):")
    print("Path:", path)
    print("Goal Reached!")

def bfs(start, goal):
    queue = deque([(start, [start])])
    visited = set()

    while queue:
        current, path = queue.popleft()

        if current == goal:
            print("\nBFS (Global Search):")
            print("Path:", path)
            print("Goal Reached!")
            return

        if current not in visited:
            visited.add(current)

            neighbors = [current - 1, current + 1]

            for neighbor in neighbors:
                if neighbor not in visited:
                    queue.append((neighbor, path + [neighbor]))

    print("Goal Not Found")

start = int(input("Enter initial state: "))
goal = int(input("Enter goal state: "))

print("\nInitial State:", start)
print("Goal State:", goal)

hill_climbing(start, goal)

bfs(start, goal)

```

```python
#3.Simulated Annealing Search
import random
import math

def heuristic(state, goal):
    return abs(goal - state)

def simulated_annealing(start, goal):
    current = start
    temperature = 100
    cooling_rate = 0.95

    path = [current]

    while temperature > 0.1:

        if current == goal:
            break

        neighbors = [current - 1, current + 1]
        next_state = random.choice(neighbors)

        current_cost = heuristic(current, goal)
        next_cost = heuristic(next_state, goal)

        delta = current_cost - next_cost

        if delta > 0:
            current = next_state
        else:
            probability = math.exp(delta / temperature)

            if random.random() < probability:
                current = next_state

        path.append(current)

        temperature *= cooling_rate

    print("\nSimulated Annealing Search:")
    print("Path:", path)

    if current == goal:
        print("Goal Reached!")
    else:
        print("Goal Not Reached")

start = int(input("Enter initial state: "))
goal = int(input("Enter goal state: "))

print("\nInitial State:", start)
print("Goal State:", goal)

simulated_annealing(start, goal)
```

```python
#4.Local Beam Search
import random
def heuristic(state, goal):
    return abs(goal - state)

def local_beam_search(start, goal, k):
    states = [start]

    print("\nLocal Beam Search:")

    while True:
        print("Current States:", states)

        if goal in states:
            print("Goal Reached!")
            return

        successors = []

        for state in states:
            neighbors = [state - 1, state + 1]

            for neighbor in neighbors:
                successors.append(neighbor)

        states = sorted(successors, key=lambda x: heuristic(x, goal))[:k]

start = int(input("Enter initial state: "))
goal = int(input("Enter goal state: "))
k = int(input("Enter beam width (k): "))

print("\nInitial State:", start)
print("Goal State:", goal)
print("Beam Width:", k)

local_beam_search(start, goal, k)
```

```python
#Category-4 1. Breadth-First Search(BFS)
from collections import deque

def bfs(graph, start, goal):
    queue = deque([[start]])
    visited = set()

    while queue:
        path = queue.popleft()
        node = path[-1]

        if node == goal:
            return path

        if node not in visited:
            visited.add(node)

            for neighbor in graph[node]:
                new_path = list(path)
                new_path.append(neighbor)
                queue.append(new_path)

    return None


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

print(bfs(graph, 'A', 'F'))
```

```python
#2.Depth-First Search (DFS)
def dfs(graph, node, goal, visited=None):
    if visited is None:
        visited = set()

    if node == goal:
        return [node]

    visited.add(node)

    for neighbor in graph[node]:
        if neighbor not in visited:
            path = dfs(graph, neighbor, goal, visited)

            if path:
                return [node] + path

    return None


print(dfs(graph, 'A', 'F'))
```

```python
#3.Depth-Limited Search (DLS)
def depth_limited_search(graph, node, goal, limit):

    if node == goal:
        return [node]

    if limit == 0:
        return None

    for neighbor in graph[node]:
        path = depth_limited_search(
            graph,
            neighbor,
            goal,
            limit - 1
        )

        if path:
            return [node] + path

    return None


print(depth_limited_search(graph, 'A', 'F', 2))
```

```python
#4.Bidirectional Search
from collections import deque

def bidirectional_search(graph, start, goal):
    if start == goal:
        return [start]

    queue1 = deque([[start]])
    queue2 = deque([[goal]])

    visited1 = {start}
    visited2 = {goal}

    while queue1 and queue2:
        path1 = queue1.popleft()
        node1 = path1[-1]

        for neighbor in graph[node1]:
            if neighbor in visited2:
                return path1 + [neighbor]

            if neighbor not in visited1:
                visited1.add(neighbor)
                queue1.append(path1 + [neighbor])

        path2 = queue2.popleft()
        node2 = path2[-1]

        for neighbor in graph[node2]:
            if neighbor in visited1:
                return path2 + [neighbor]

            if neighbor not in visited2:
                visited2.add(neighbor)
                queue2.append(path2 + [neighbor])

    return None
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'G'],
    'F': ['C'],
    'G': ['E']
}

print(bidirectional_search(graph, 'A', 'G'))
```
